## Trees
How to build one in 3 simple steps:

Make a "Box" (Node)
Every spot on a tree is just a box that holds two things:

A value (like a name or number).
"Hands" or links pointing to other boxes underneath it (its children).

Pick a Starting Point (The Root)
Create your very first box. This is the trunk/root where everything starts.

Attach Branches (Connect Children)
Create more boxes and link them to the first box. Those new boxes can have their own children, and so on.

```python
class Tree:
    def __init__(self, val):
        self.parent = None
        self.childrens = []
        self.val = val
    
    def add_child(self, child):
        child.parent = self
        self.childrens.append(child)

    def get_level(self):
        curr_parent = self.parent
        level = 0
        while curr_parent:
            level += 1
            curr_parent = curr_parent.parent
        return level 

    def print_tree(self,):
        print(f"{self.get_level()*3*' '}|__{self.val}")
        for child in self.childrens:
            child.print_tree()       
```

```python
root = Tree(1)
child1 = Tree(2)
child2 = Tree(3)
root.add_child(child1)
root.add_child(child2)

child4 = Tree(4)
child5 = Tree(5)

child1.add_child(child4)
child1.add_child(child5)
root.print_tree()

child5.get_level()
```

## 🌳 Binary Tree & BST (Quick Notes)

### 1. Binary Tree
* *"Binary"* means **two**.
* Each node can have **at most 2 children** (0, 1, or 2) — called **Left** and **Right**.

---

### 2. Binary Search Tree (BST)
A binary tree arranged with one simple rule:
> **`Left <= Parent <= Right`**
* **Left side:** Contains smaller (or equal) values.
* **Right side:** Contains larger (or equal) values.
* *Why?* Makes searching super fast like a "Higher or Lower" game.

---

### 3. Basic Operations

* 🔍 **Search:**  
  * Smaller than current? Go **Left**.  
  * Bigger than current? Go **Right**.  
  * Equal? **Found it!**

* ➕ **Insert:**  
  * Follow the Search rule until you hit an empty spot, then attach the new node there.

* ❌ **Delete:**  
  * **0 children (Leaf):** Just delete it.  
  * **1 child:** The child takes the parent's place.  
  * **2 children:** Replace the node with the smallest value from its right subtree, then remove that duplicate.


```python
def print_tree(node, prefix="", is_tail=True, label="Root"):
    """
    Prints a visual representation of a binary tree.
    """
    if node is None:
        return
    # Print the current node with appropriate branch styling
    connector = "└── " if is_tail else "├── "
    print(f"{prefix}{connector}[{label}] {node.val}")
    # Prepare prefix for child nodes
    new_prefix = prefix + ("    " if is_tail else "│   ")
    # Collect existing children
    children = []
    if node.left or node.right:
        # Include None place-holders if one child is missing to preserve structure
        children.append((node.left, "L"))
        children.append((node.right, "R"))
    # Recursively print children
    for i, (child, child_label) in enumerate(children):
        is_last = (i == len(children) - 1)
        if child:
            print_tree(child, new_prefix, is_last, child_label)
        else:
            connector_child = "└── " if is_last else "├── "
            print(f"{new_prefix}{connector_child}[{child_label}] None")
```

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

class BinaryTree:
    def __init__(self,root):
        self.root = root

    def insert(self, val):
        new_node = Node(val)
        node = self.root
        parent = node
        while node is not None:
            parent = node
            if val > node.val:
                node = node.right
            elif val < node.val:
                node = node.left
        if val > parent.val:
            parent.right = new_node
        elif val < parent.val:
            parent.left = new_node

    def delete(self, node, val):
        if val > node.val:
            node.right = self.delete(node.right, val)
        elif val < node.val:
            node.left = self.delete(node.left, val)
        else:
            if not node.left:
                return node.right
            if not node.right:
                return node.left
            successor = node.right
            while successor.left:
                successor = successor.left
            node.val = successor.val 
            node.right = self.delete(node.right, successor.val)
        return node                    

    def search(self, val):
        node = root
        level = 0
        while node:
            level += 1
            if val > node.val:
                node = node.right
            elif val < node.val:
                node = node.left
            if val == node.val:
                break

        if not node:
            print(f"{val} not found in Tree")   
        else:     
            print(f"{val} found in Tree at Level {level}")                    
```

```python
class Traversing_With_Recursion:
    def inorder(self,node):
        if not node:
            return
        self.inorder(node.left)    
        print(f"{node.val}->",end="")
        self.inorder(node.right)

    def preorder(self,node):
        if not node:
            return
        print(f"{node.val}->",end="")    
        self.preorder(node.left)    
        self.preorder(node.right)

    def postorder(self,node):
        if not node:
            return
        self.postorder(node.left)    
        self.postorder(node.right) 
        print(f"{node.val}->",end="")     
```

## 🕸️ Graphs (Quick Notes)

A **Graph** is just a network of connected points — like cities connected by roads, or people on social media.

---

### 1. Vertices (Nodes)
* **What it is:** The **dots** in the network.
* **Example:** People (Alice, Bob) or Cities (Mumbai, Delhi).

### 2. Edges
* **What it is:** The **lines** that connect two vertices.
* **Example:** A friendship between two people, or a highway between two cities.
* *Can be:*
  * **Undirected:** Two-way street (Facebook friends).
  * **Directed:** One-way arrow (Instagram follower: A follows B, but B might not follow A).

  (A) -------- (B) <-- Vertices: A, B, C \ / <-- Edges: The lines connecting them \ / (C)

---

### 3. Graph Traversing (Visiting Every Vertex)

Visiting all vertices without getting stuck in a loop requires a strategy:

#### 1) 🌊 BFS (Breadth-First Search) — *Level by Level*
* **How it works:** Visits all your **immediate neighbors first**, then their neighbors next (like ripples expanding in water).
* **Analogy:** Checking all your closest friends before meeting friends-of-friends.
* **Tool used:** **Queue** (First In, First Out).
* **Best for:** Finding the **shortest path**.

#### 2) 🕳️ DFS (Depth-First Search) — *Down the Rabbit Hole*
* **How it works:** Picks a path and goes **as deep as possible** until it hits a dead end, then backtracks.
* **Analogy:** Exploring a maze by following one wall until you can't go further, then turning back.
* **Tool used:** **Stack** or **Recursion** (Last In, First Out).
* **Best for:** Solving puzzles, mazes, and cycle detection.



```python
class Graph:
    def __init__(self,graph):
        self.graph = graph
        self.visited = [False for index in graph]
        self.bfs_list = []
        self.dfs_list = []

    def bfs(self,node):
        from collections import deque
        q = deque([node])
        self.visited[node] = True
        while q:
            node = q.popleft()
            self.bfs_list.append(node)
            for neighbour in self.graph[node]:
                if not self.visited[neighbour]:
                    q.append(neighbour)
                    self.visited[neighbour] = True

    def dfs(self,node):
        self.dfs_list.append(node)
        self.visited[node] = True
        for neighbour in self.graph[node]:
            if not self.visited[neighbour]:
               self.dfs(neighbour)

    def dfs_iterative(self,node):
        stack = [node]
        self.visited[node] = True
        while stack:
            node = stack.pop()
            self.dfs_list.append(node)
            for neighbour in self.graph[node]:
                if not self.visited[neighbour]:
                    stack.append(neighbour)
                    self.visited[neighbour] = True
```

```python
graph = [[1,2], [0,4,2,3], [0,1,3], [1,2], [1]]
graph_obj = Graph(graph)
graph_obj.dfs_iterative(0)
graph_obj.dfs_list
```
