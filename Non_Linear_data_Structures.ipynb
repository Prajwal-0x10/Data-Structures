## 1. General Trees (N-ary Tree)
**Core Concept (Gist):** 
A collection of nodes connected by directed edges, originating from a single **Root** node. Every node can have **0 or more children** and exactly one parent (except the root).

### 🔑 Key Terminology:
- **Root**: Topmost node without a parent.
- **Child / Parent**: Nodes linked directly below / above each other.
- **Leaf Node**: A node with 0 children.
- **Level / Depth**: Number of edges from the root to the node (Root is at Level 0).
- **Height**: Maximum number of edges from root to the furthest leaf.

### ⏱️ Complexity Summary:
| Operation | Time Complexity | Space Complexity |
| :--- | :---: | :---: |
| **Add Child** | $O(1)$ | $O(1)$ |
| **Get Level / Depth** | $O(\text{depth})$ | $O(1)$ |
| **Print / Traverse Tree** | $O(N)$ | $O(N)$ (recursion stack) |



````python
class Tree:
    def __init__(self, val):
        self.parent = None
        self.childrens = []
        self.val = val
    
    def add_child(self, child):
        child.parent = self
        self.childrens.append(child)

    def get_level(self):
        curr_parent = self.parent
        level = 0
        while curr_parent:
            level += 1
            curr_parent = curr_parent.parent
        return level 

    def print_tree(self,):
        print(f"{self.get_level()*3*' '}|__{self.val}")
        for child in self.childrens:
            child.print_tree()       
````

````python
root = Tree(1)
child1 = Tree(2)
child2 = Tree(3)
root.add_child(child1)
root.add_child(child2)

child4 = Tree(4)
child5 = Tree(5)

child1.add_child(child4)
child1.add_child(child5)
root.print_tree()

child5.get_level()
````

## 2. Binary Tree & Binary Search Tree (BST)
**Core Concept (Gist):**
- **Binary Tree**: A tree where every node has **at most 2 children** (Left and Right).
- **Binary Search Tree (BST)**: A binary tree satisfying the **BST Invariant**:
  $$\text{Left Subtree Values} < \text{Parent Node Value} < \text{Right Subtree Values}$$

```
        [Root] 5
       ┌───┴───┐
     [L] 2   [R] 8
       └──┐   ┌──┘
        [R] 4 [L] 6
        ┌──┘
      [L] 3
```

### ⚙️ Core BST Operations:
| Operation | Average Case | Worst Case (Skewed) | Description |
| :--- | :---: | :---: | :--- |
| **Search** | $O(\log N)$ | $O(N)$ | Halves the search space at each step by comparing with current node. |
| **Insert** | $O(\log N)$ | $O(N)$ | Traverses down the tree according to BST property and attaches as a leaf. |
| **Delete** | $O(\log N)$ | $O(N)$ | Requires handling 3 distinct cases (Leaf, 1 Child, 2 Children). |

### 💡 The 3 Cases of BST Deletion:
1. **Case 1 (Leaf Node):** Node has no children $\to$ Simply remove it (`return None`).
2. **Case 2 (One Child):** Node has one child $\to$ Replace the node with its child.
3. **Case 3 (Two Children):** Node has two children $\to$ Find the **In-Order Successor** (smallest node in the right subtree), copy its value to the current node, and recursively delete the successor from the right subtree.



````python
def print_tree(node, prefix="", is_tail=True, label="Root"):
    """
    Prints a visual representation of a binary tree.
    """
    if node is None:
        return
    # Print the current node with appropriate branch styling
    connector = "└── " if is_tail else "├── "
    print(f"{prefix}{connector}[{label}] {node.val}")
    # Prepare prefix for child nodes
    new_prefix = prefix + ("    " if is_tail else "│   ")
    # Collect existing children
    children = []
    if node.left or node.right:
        # Include None place-holders if one child is missing to preserve structure
        children.append((node.left, "L"))
        children.append((node.right, "R"))
    # Recursively print children
    for i, (child, child_label) in enumerate(children):
        is_last = (i == len(children) - 1)
        if child:
            print_tree(child, new_prefix, is_last, child_label)
        else:
            connector_child = "└── " if is_last else "├── "
            print(f"{new_prefix}{connector_child}[{child_label}] None")
````

````python
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

class BinaryTree:
    def __init__(self,root):
        self.root = root

    def insert(self, val):
        new_node = Node(val)
        node = self.root
        parent = node
        while node is not None:
            parent = node
            if val > node.val:
                node = node.right
            elif val < node.val:
                node = node.left
        if val > parent.val:
            parent.right = new_node
        elif val < parent.val:
            parent.left = new_node

    def delete(self, node, val):
        if val > node.val:
            node.right = self.delete(node.right, val)
        elif val < node.val:
            node.left = self.delete(node.left, val)
        else:
            if not node.left:
                return node.right
            if not node.right:
                return node.left
            successor = node.right
            while successor.left:
                successor = successor.left
            node.val = successor.val 
            node.right = self.delete(node.right, successor.val)
        return node                    

    def search(self, val):
        node = root
        level = 0
        while node:
            level += 1
            if val > node.val:
                node = node.right
            elif val < node.val:
                node = node.left
            if val == node.val:
                break

        if not node:
            print(f"{val} not found in Tree")   
        else:     
            print(f"{val} found in Tree at Level {level}")                    
````

````python
root = Node(5)
tree = BinaryTree(root)
tree.insert(2)
tree.insert(4)
tree.insert(7)
tree.insert(6)
tree.insert(8)
tree.insert(3)

root = tree.delete(tree.root, 7)

print_tree(root)
````

## 3. Tree Traversals — Recursive Approach (DFS)
**Core Concept (Gist):**
Traversal is the process of visiting all nodes in a tree exactly once. With recursion, the system call stack tracks our backtrack path.

### 🔄 The 3 Classical DFS Orders:
| Traversal Type | Visit Order | Mental Model / Typical Use Case |
| :--- | :---: | :--- |
| **In-Order** | $\text{Left} \to \text{Root} \to \text{Right}$ | Returns elements of a BST in **sorted ascending order**. |
| **Pre-Order** | $\text{Root} \to \text{Left} \to \text{Right}$ | Great for **copying / serializing** tree structure. |
| **Post-Order** | $\text{Left} \to \text{Right} \to \text{Root}$ | Great for **bottom-up deletion** (freeing children before parent) & evaluating math expression trees. |

- **Time Complexity:** $O(N)$ (every node visited once)
- **Space Complexity:** $O(h)$ where $h$ is tree height (recursion call stack)



````python
class Traversing_With_Recursion:
    def inorder(self,node):
        if not node:
            return
        self.inorder(node.left)    
        print(f"{node.val}->",end="")
        self.inorder(node.right)

    def preorder(self,node):
        if not node:
            return
        print(f"{node.val}->",end="")    
        self.preorder(node.left)    
        self.preorder(node.right)

    def postorder(self,node):
        if not node:
            return
        self.postorder(node.left)    
        self.postorder(node.right) 
        print(f"{node.val}->",end="")     
````

````python
traverse_obj = Traversing_With_Recursion()
traverse_obj.inorder(root)
print()
traverse_obj.preorder(root)
print()
traverse_obj.postorder(root)
````

## 4. Tree Traversals — Iterative Approach (No Recursion)
**Core Concept (Gist):**
Recursive solutions can hit recursion depth limits on deep or skewed trees. Iterative traversals eliminate this by managing our own explicit data structures (**Queue** or **Stack**).

### 📊 Traversal Strategies:
1. **Breadth-First Search (BFS / Level-Order):**
   - Uses a **FIFO Queue** (`collections.deque`).
   - Visits nodes level-by-level, left-to-right.
   - **Complexity:** Time $O(N)$ | Space $O(W)$ (where $W$ is maximum width of tree).
2. **Iterative DFS (In-Order, Pre-Order, Post-Order):**
   - Uses a **LIFO Stack** to simulate the recursion call stack.
   - **In-Order Trick:** Walk all the way down left branches pushing to stack; pop, visit, and shift to right child.
   - **Post-Order Trick:** Traverse in reverse pre-order ($\text{Root} \to \text{Right} \to \text{Left}$) and reverse the output list at the end!



````python
from collections import deque
class Traverse_Without_Recursion:
    def bfs(self,node):
        q = deque([node])
        while q:
            node = q.popleft()
            print(f"{node.val}->", end = "")
            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)

    def preorder(self,node):
        s = [node]
        while s:
            node = s.pop()
            print(f"{node.val}->", end = "")
            if node.right:
                s.append(node.right)
            if node.left:
                s.append(node.left)

    def postorder(self,node):
        s = [node]
        pre_order = []
        while s:
            node = s.pop()
            pre_order.append(str(node.val))
            if node.left:
                s.append(node.left)
            if node.right:
                s.append(node.right)
        print("->".join(pre_order[::-1]))

    def inorder(self, node):
        s = []
        curr = node
        while s or curr:
            while curr:
                s.append(curr) 
                curr = curr.left
            curr = s.pop()
            print(curr.val,end="->")
            curr = curr.right             
                        

````

## 5. Graphs: Adjacency List & Traversal Algorithms
**Core Concept (Gist):**
A **Graph** $G = (V, E)$ is a non-linear network of **Vertices (Nodes)** connected by **Edges (Links)**. Unlike trees:
- Graphs can contain **cycles** (circular paths).
- Graphs do not require a single root node.
- Can be **directed / undirected**, and **weighted / unweighted**.

### 🕸️ Representation: Adjacency List
An adjacency list represents the graph as a list or dictionary where index $i$ holds all neighbors of vertex $i$.
- **Space Complexity:** $O(V + E)$ (highly memory-efficient for sparse graphs).

```
   Adjacency List Graph:
   0: [1, 2]
   1: [0, 4, 2, 3]
   2: [0, 1, 3]
   3: [1, 2]
   4: [1]

       0 ───── 1 ───── 4
       │     ╱ │
       │   ╱   │
       2 ───── 3
```

### 🚀 BFS vs. DFS:
| Algorithm | Auxiliary DS | Exploration Strategy | Time Complexity | Space Complexity | Best Used For |
| :--- | :---: | :--- | :---: | :---: | :--- |
| **BFS (Breadth-First)** | **Queue** (`deque`) | Explores neighbors layer-by-layer | $O(V + E)$ | $O(V)$ | **Shortest path** in unweighted graphs, peer-to-peer discovery. |
| **DFS (Depth-First)** | **Stack** / Recursion | Explores as deep as possible before backtracking | $O(V + E)$ | $O(V)$ | **Cycle detection**, topological sorting, maze solving. |

> **⚠️ Essential Rule:** Always track `visited` nodes (via a boolean list or hash set) to prevent infinite loops in cyclic graphs!



````python
class Graph:
    def __init__(self,graph):
        self.graph = graph
        self.visited = [False for index in graph]
        self.bfs_list = []
        self.dfs_list = []

    def bfs(self,node):
        from collections import deque
        q = deque([node])
        self.visited[node] = True
        while q:
            node = q.popleft()
            self.bfs_list.append(node)
            for neighbour in self.graph[node]:
                if not self.visited[neighbour]:
                    q.append(neighbour)
                    self.visited[neighbour] = True

    def dfs(self,node):
        self.dfs_list.append(node)
        self.visited[node] = True
        for neighbour in self.graph[node]:
            if not self.visited[neighbour]:
               self.dfs(neighbour)

    def dfs_iterative(self,node):
        stack = [node]
        self.visited[node] = True
        while stack:
            node = stack.pop()
            self.dfs_list.append(node)
            for neighbour in self.graph[node]:
                if not self.visited[neighbour]:
                    stack.append(neighbour)
                    self.visited[neighbour] = True
````

````python
graph = [[1,2], [0,4,2,3], [0,1,3], [1,2], [1]]
graph_obj = Graph(graph)
graph_obj.dfs_iterative(0)
graph_obj.dfs_list
````

````python
graph = [[1,2], [0,4,2,3], [0,1,3], [1,2], [1]]
graph_obj = Graph(graph)
graph_obj.bfs(0)
graph_obj.bfs_list
````

````python
graph = [[1,2], [0,4,2,3], [0,1,3], [1,2], [1]]
graph_obj = Graph(graph)
graph_obj.dfs(0)
graph_obj.dfs_list
````

## 📌 Summary & Quick Reference Cheat Sheet

| Structure / Algorithm | Core Mechanism | Time Complexity | Auxiliary Space | Best Used When |
| :--- | :--- | :---: | :---: | :--- |
| **N-ary Tree** | Multi-child hierarchy | $O(1)$ child add | $O(N)$ storage | File systems, organizational hierarchies, DOM trees |
| **Binary Search Tree** | $\text{Left} < \text{Root} < \text{Right}$ | $O(\log N)$ avg | $O(N)$ | Fast search, dynamic sorted data retrieval |
| **Tree BFS (Level-Order)**| FIFO Queue | $O(N)$ | $O(W)$ width | Level-by-level inspection, shortest path in tree |
| **Tree DFS (In/Pre/Post)**| LIFO Stack / Recursion | $O(N)$ | $O(h)$ height | Hierarchical evaluation, sorted dump (In-order) |
| **Graph BFS** | Queue + `visited` | $O(V + E)$ | $O(V)$ | Shortest path in unweighted networks |
| **Graph DFS** | Stack/Recursion + `visited` | $O(V + E)$ | $O(V)$ | Exhaustive search, cycle detection, path existence |
