## 🔢 Sorting Algorithms (Quick Revision)

Sorting is just putting items in order (usually smallest to largest). Here is how each algorithm thinks:

---

### 1. 🫧 Bubble Sort — *"Bubbling up the heavy elements"*
* **How it works:** Compares side-by-side neighbors. If the left one is bigger, swap them. Repeat until no more swaps are needed.
* **Real-life analogy:** Heavy bubbles float to the top; with each pass, the biggest number walks to the end of the line.
* **Speed:** Slow — $O(n^2)$

---

### 2. 🎯 Selection Sort — *"Pick the smallest"*
* **How it works:** Scans the entire unsorted list, finds the absolute **smallest** item, and swaps it with the front. Then finds the next smallest, and so on.
* **Real-life analogy:** Scanning a group of people, picking the shortest person to stand first in line, then repeating.
* **Speed:** Slow — $O(n^2)$

---

### 3. 🃏 Insertion Sort — *"Sorting playing cards"*
* **How it works:** Takes one item at a time from the unsorted side and **inserts** it into its correct spot among the already-sorted items.
* **Real-life analogy:** How you organize playing cards in your hand as they are dealt to you one by one.
* **Speed:** $O(n^2)$ normally, but **super fast** $O(n)$ if the list is already almost sorted!

---

### 4. 🪓 Merge Sort — *"Divide and Conquer"*
* **How it works:** 
  1. Keeps cutting the list in half until every item is alone.
  2. Merges small pieces back together in sorted pairs.
* **Real-life analogy:** Tearing a stack of papers in half until you have single sheets, then neatly zippering them back together in order.
* **Speed:** Fast & Consistent — $O(n \log n)$

---

### 5. ⚡ Quick Sort — *"Pick a leader (Pivot)"*
* **How it works:** 
  1. Picks one number as the **pivot** (benchmark).
  2. Moves everything smaller to its left, and everything bigger to its right.
  3. Repeats the same process for left and right sides.
* **Real-life analogy:** A teacher says: *"Everyone shorter than me to my left, everyone taller to my right!"*
* **Speed:** Very fast on average — $O(n \log n)$

---

### 6. 🏔️ Heap Sort — *"King of the Hill"*
* **How it works:** Organizes numbers into a binary heap (a pyramid where the biggest number is always at the very top). Pops the top number out, re-balances the heap, and repeats.
* **Real-life analogy:** King of the Hill — the winner takes the crown at the top; once crowned, they step aside and the next battle decides the next king.
* **Speed:** Fast & Predictable — $O(n \log n)$

---

### ⚡ Quick Cheat Sheet

| Algorithm | Real-Life Idea | Average Speed |
| :--- | :--- | :--- |
| **Bubble** | Swap adjacent neighbors | $O(n^2)$ (Slow) |
| **Selection** | Select the minimum, move to front | $O(n^2)$ (Slow) |
| **Insertion** | Insert like sorting cards in hand | $O(n^2)$ (Fast if nearly sorted) |
| **Merge** | Split in halves, zipper back together | $O(n \log n)$ (Reliable) |
| **Quick** | Partition around a pivot | $O(n \log n)$ (Usually fastest) |
| **Heap** | Pull max from a pyramid/heap | $O(n \log n)$ (Consistent) |


```python
class Sorting:
    def __init__(self, array):
        self.array = array
        self.length = len(array)
    
    def swap(self,first, second):
        self.array[first], self.array[second] = self.array[second], self.array[first]

    def Bubble_Sort(self,):
        for start in range(0,self.length):
            for iteration in range(start,self.length):
                if self.array[iteration-1] > self.array[iteration]:
                    self.swap(iteration-1, iteration)   

    def Selection_Sort(self,):
        for start in range(0, self.length):
            min_index = start
            for index in range(start,self.length):
                if self.array[index] < self.array[min_index]:
                    min_index = index
            if min_index != start:
                self.swap(min_index, start)  

    def Insertion_Sort(self,):
        for start in range(0, self.length):
            element = self.array[start]
            index = start
            while index > 0 and element < self.array[index-1]:
                self.array[index] = self.array[index-1]
                index -= 1
            self.array[index] = element    

    def Merge_Sort(self,l,h):
        
        def merge(l,mid,h):
            temp_list = []
            first = l
            second = mid + 1
            while first <= mid and second <= h:
                if self.array[first] > self.array[second]:
                    temp_list.append(self.array[second])
                    second += 1
                else:
                    temp_list.append(self.array[first])    
                    first += 1        
            while first <= mid:
                temp_list.append(self.array[first]) 
                first += 1
            while second <= h:
                temp_list.append(self.array[second]) 
                second += 1
            
            for index in range(l,h+1):
                self.array[index] = temp_list[index-l] 
            #print(self.array)               

        if l < h:
            mid = (l+h)//2
            self.Merge_Sort(l, mid)
            self.Merge_Sort(mid+1, h)
            merge(l,mid,h)  


    def Quick_Sort(self, l, h):
        def get_pivot(l,h):
            start = l
            end = h-1
            pivot_element = self.array[start]
            while start < end:
                while start < h and self.array[start] < pivot_element:
                    start += 1
                while end > l and self.array[end] >= pivot_element:
                    end -= 1
                if start < end:
                    self.swap(start, end)        
            self.swap(l, end)
            return end    

        if l < h:
            pivot = get_pivot(l, h)
            self.Quick_Sort(l, pivot)
            self.Quick_Sort(pivot+1, h)
                                                       
```

```python
array = [10,5,23,5,78,6,43,7]
sort_obj = Sorting(array)
sort_obj.Bubble_Sort()
sort_obj.Selection_Sort()
sort_obj.Insertion_Sort()
sort_obj.Merge_Sort(0,len(array)-1)
sort_obj.Quick_Sort(0,len(array))
print(sort_obj.array)
```

## 🏔️ Understanding Heaps (Before Heap Sort)

Before we can sort using **Heap Sort**, we need to understand what a **Heap** is and how it functions.

---

### 1. What is a Heap?
A **Heap** is a complete binary tree (filled from left to right) that follows a strict **hierarchy rule**:

* **Max-Heap:** Every parent is **greater than or equal to** its children.  
  👉 *The largest number is always at the very top (root).*
* **Min-Heap:** Every parent is **smaller than or equal to** its children.  
  👉 *The smallest number is always at the very top (root).*

  Max-Heap Example:
       [ 50 ]        <-- Boss (Largest element at root)
      /      \
  [ 30 ]    [ 20 ]   <-- Must be smaller than 50
  /    \
[10]   [15]          <-- Must be smaller than 30

> **Cool trick:** Even though we draw it like a tree, a heap is usually stored as a simple **Python list `[50, 30, 20, 10, 15]`**!  
> For any index `i`:
> * **Left child:** `2 * i + 1`
> * **Right child:** `2 * i + 2`
> * **Parent:** `(i - 1) // 2`

---

### 2. Core Heap Operations

#### a) 🛠️ `Heapify` (Fixing the pyramid)
* **What it does:** If one node violates the rule (e.g., a small number is sitting as the parent of big numbers), `heapify` swaps it with its biggest child.
* **Analogy:** Sinking a stone in water — the small number "sinks down" until it reaches its correct level.

#### b) ➕ `Insert` (Adding an element)
1. Drop the new number at the **very end** (bottom-most, right-most spot).
2. Compare it with its parent: if it's bigger, **bubble it up** (swap with parent).
3. Repeat until the heap rule is satisfied.

#### c) 👑 `Extract Max` (Removing the top element)
1. Take the **root** (the largest number) out.
2. Put the **last element** of the heap into the root spot.
3. Call `heapify` on the root so this new element sinks down to its right place.

#### d) 🏗️ `Build Heap`
* Takes any random, messy list of numbers and organizes them into a valid Max-Heap from the bottom up.

---

```python
class MaxHeap:

    def __init__(self,):
        self.heap = []

    def swap(self,first, second):
        self.heap[first], self.heap[second] = self.heap[second], self.heap[first]
    
    def print_heap(self,):
        print(self.heap)

    def insert(self, val):
        self.heap.append(val)
        index = len(self.heap)-1
        parent = (index-1)//2   
        while parent >=0 and self.heap[parent] < self.heap[index]:
            self.swap(parent, index)
            index = parent
            parent = (index-1)//2

    def delete_element(self,):
        # IN heap we always delete only root element       
        length = len(self.heap)
        self.swap(length-1,0)
        element = self.heap.pop()
        length = len(self.heap)
        print(f"{element} removed from heap")
        index = 0
        while True:
            left = 2*index+1
            right = 2*index+2
            large = index
            if left < length and self.heap[left] >= self.heap[large]:
               large = left
            if right < length and self.heap[right] > self.heap[large]:
                large = right
            if large == index:
                break
            self.swap(large,index)
            index = large    
        return element       

            
```

```python
heap_obj = MaxHeap()
heap_obj.insert(5)
heap_obj.insert(4)
heap_obj.insert(3)
heap_obj.insert(2)
heap_obj.insert(1)
heap_obj.insert(7)
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
```

```python
class MinHeap:

    def __init__(self,):
        self.heap = []

    def swap(self,first, second):
        self.heap[first], self.heap[second] = self.heap[second], self.heap[first]
    
    def print_heap(self,):
        print(self.heap)

    def insert(self, values):
        for val in values:
            self.heap.append(val)
            index = len(self.heap)-1
            parent = (index-1)//2   
            while parent >=0 and self.heap[parent] > self.heap[index]:
                self.swap(parent, index)
                index = parent
                parent = (index-1)//2

    def delete_element(self,):
        # IN heap we always delete only root element 
        if not self.heap: return      
        length = len(self.heap)
        self.swap(length-1,0)
        element = self.heap.pop()
        length = len(self.heap)
        print(f"{element} removed from heap")
        index = 0
        while True:
            left = 2*index+1
            right = 2*index+2
            large = index
            if left < length and self.heap[left] <= self.heap[large]:
               large = left
            if right < length and self.heap[right] < self.heap[large]:
                large = right
            if large == index:
                break
            self.swap(large,index)
            index = large    
        return element  
```

```python
heap_obj = MinHeap()
heap_obj.insert([5])
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
```


### 3. How this leads to Heap Sort!

Once you understand these operations, **Heap Sort** is basically just a 3-step loop:

1. **Build a Max-Heap** from the input list (biggest item is now at index `0`).
2. **Swap** index `0` (the biggest) with the **last element** (it's now in its final sorted position at the end).
3. **Shrink** the heap by 1 and call **`heapify`** on the top to find the next biggest number.
4. Repeat until the whole list is sorted!


```python
heap_obj = MinHeap()
heap_obj.insert([9,8,7,6,5,4,3,2,1])
heap_obj.print_heap()
sorted_array = []
while heap_obj.heap:
    element = heap_obj.delete_element()
    sorted_array.append(element)
print(sorted_array)    
```

```python
class Heap_Sort:
    def __init__(self,heap=None):
        self.heap = [] if not heap else heap
    
    def build_heap(self,n):
        for i in range(n//2-1, -1, -1):
            self.heapify(i,n)

    def heapify(self, start, end):
        while True:
            left = 2*start+1
            right = 2*start+2
            large = start
            if left < end and self.heap[left] > self.heap[large]:
                large = left
            if right < end and self.heap[right] > self.heap[large]:
                large = right
            if large == start:
                break
            self.heap[start], self.heap[large] = self.heap[large], self.heap[start]
            start = large 

    def Sort(self,):
        for i in range(len(self.heap)-1,-1,-1):
            self.heap[0], self.heap[i] = self.heap[i], self.heap[0]
            self.heapify(0,i)                
```

```python
heap = [4,10,3,5,1,11]
heap_obj = Heap_Sort(heap)
heap_obj.build_heap(len(heap))
heap_obj.Sort()
heap_obj.heap
```
