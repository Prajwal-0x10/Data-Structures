## Sorting
1) Bubble Sort
2) Selction Sort
3) Insertion Sort
4) Merge Sort
5) Quick Sort
6) Heap Sort

```python
class Sorting:
    def __init__(self, array):
        self.array = array
        self.length = len(array)
    
    def swap(self,first, second):
        self.array[first], self.array[second] = self.array[second], self.array[first]

    def Bubble_Sort(self,):
        for start in range(0,self.length):
            for iteration in range(start,self.length):
                if self.array[iteration-1] > self.array[iteration]:
                    self.swap(iteration-1, iteration)   

    def Selection_Sort(self,):
        for start in range(0, self.length):
            min_index = start
            for index in range(start,self.length):
                if self.array[index] < self.array[min_index]:
                    min_index = index
            if min_index != start:
                self.swap(min_index, start)  

    def Insertion_Sort(self,):
        for start in range(0, self.length):
            element = self.array[start]
            index = start
            while index > 0 and element < self.array[index-1]:
                self.array[index] = self.array[index-1]
                index -= 1
            self.array[index] = element    

    def Merge_Sort(self,l,h):
        
        def merge(l,mid,h):
            temp_list = []
            first = l
            second = mid + 1
            while first <= mid and second <= h:
                if self.array[first] > self.array[second]:
                    temp_list.append(self.array[second])
                    second += 1
                else:
                    temp_list.append(self.array[first])    
                    first += 1        
            while first <= mid:
                temp_list.append(self.array[first]) 
                first += 1
            while second <= h:
                temp_list.append(self.array[second]) 
                second += 1
            
            for index in range(l,h+1):
                self.array[index] = temp_list[index-l] 
            #print(self.array)               

        if l < h:
            mid = (l+h)//2
            self.Merge_Sort(l, mid)
            self.Merge_Sort(mid+1, h)
            merge(l,mid,h)  


    def Quick_Sort(self, l, h):
        def get_pivot(l,h):
            start = l
            end = h-1
            pivot_element = self.array[start]
            while start < end:
                while start < h and self.array[start] < pivot_element:
                    start += 1
                while end > l and self.array[end] >= pivot_element:
                    end -= 1
                if start < end:
                    self.swap(start, end)        
            self.swap(l, end)
            return end    

        if l < h:
            pivot = get_pivot(l, h)
            self.Quick_Sort(l, pivot)
            self.Quick_Sort(pivot+1, h)
                                                       
```

```python
array = [10,5,23,5,78,6,43,7]
sort_obj = Sorting(array)
sort_obj.Bubble_Sort()
sort_obj.Selection_Sort()
sort_obj.Insertion_Sort()
sort_obj.Merge_Sort(0,len(array)-1)
sort_obj.Quick_Sort(0,len(array))
print(sort_obj.array)
```

## Heap Sort
Before starting heap sort we must know how heap works and what are its operations


```python
class MaxHeap:

    def __init__(self,):
        self.heap = []

    def swap(self,first, second):
        self.heap[first], self.heap[second] = self.heap[second], self.heap[first]
    
    def print_heap(self,):
        print(self.heap)

    def insert(self, val):
        self.heap.append(val)
        index = len(self.heap)-1
        parent = (index-1)//2   
        while parent >=0 and self.heap[parent] < self.heap[index]:
            self.swap(parent, index)
            index = parent
            parent = (index-1)//2

    def delete_element(self,):
        # IN heap we always delete only root element       
        length = len(self.heap)
        self.swap(length-1,0)
        element = self.heap.pop()
        length = len(self.heap)
        print(f"{element} removed from heap")
        index = 0
        while True:
            left = 2*index+1
            right = 2*index+2
            large = index
            if left < length and self.heap[left] >= self.heap[large]:
               large = left
            if right < length and self.heap[right] > self.heap[large]:
                large = right
            if large == index:
                break
            self.swap(large,index)
            index = large    
        return element       

            
```

```python
heap_obj = MaxHeap()
heap_obj.insert(5)
heap_obj.insert(4)
heap_obj.insert(3)
heap_obj.insert(2)
heap_obj.insert(1)
heap_obj.insert(7)
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
```

```python
class MinHeap:

    def __init__(self,):
        self.heap = []

    def swap(self,first, second):
        self.heap[first], self.heap[second] = self.heap[second], self.heap[first]
    
    def print_heap(self,):
        print(self.heap)

    def insert(self, values):
        for val in values:
            self.heap.append(val)
            index = len(self.heap)-1
            parent = (index-1)//2   
            while parent >=0 and self.heap[parent] > self.heap[index]:
                self.swap(parent, index)
                index = parent
                parent = (index-1)//2

    def delete_element(self,):
        # IN heap we always delete only root element 
        if not self.heap: return      
        length = len(self.heap)
        self.swap(length-1,0)
        element = self.heap.pop()
        length = len(self.heap)
        print(f"{element} removed from heap")
        index = 0
        while True:
            left = 2*index+1
            right = 2*index+2
            large = index
            if left < length and self.heap[left] <= self.heap[large]:
               large = left
            if right < length and self.heap[right] < self.heap[large]:
                large = right
            if large == index:
                break
            self.swap(large,index)
            index = large    
        return element  
```

```python
heap_obj = MinHeap()
heap_obj.insert([5])
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
heap_obj.delete_element()
heap_obj.print_heap()
```

```python
class Heap_Sort:
    def __init__(self,heap=None):
        self.heap = [] if not heap else heap
    
    def build_heap(self,n):
        for i in range(n//2-1, -1, -1):
            self.heapify(i,n)

    def heapify(self, start, end):
        while True:
            left = 2*start+1
            right = 2*start+2
            large = start
            if left < end and self.heap[left] > self.heap[large]:
                large = left
            if right < end and self.heap[right] > self.heap[large]:
                large = right
            if large == start:
                break
            self.heap[start], self.heap[large] = self.heap[large], self.heap[start]
            start = large 

    def Sort(self,):
        for i in range(len(self.heap)-1,-1,-1):
            self.heap[0], self.heap[i] = self.heap[i], self.heap[0]
            self.heapify(0,i)                
```

```python
heap = [4,10,3,5,1,11]
heap_obj = Heap_Sort(heap)
heap_obj.build_heap(len(heap))
heap_obj.Sort()
heap_obj.heap
```
