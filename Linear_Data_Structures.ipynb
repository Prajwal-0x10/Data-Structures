## Arrays
Collection of Elements of same or different data types.

```python
array = [1,2,3,4,5,6]
```

Operations on Array
1) Insert at position
2) Insert at end
3) Search an element
4) Delete an element


```python
class Array:
    def __init__(self,):
        self.array = []

    def insert_at_end(self, element):
        self.array.append(element)
        self.print_array()

    def insert_at_position(self, position, element):
        self.array.insert(position, element)
        self.print_array()

    def search_an_element(self, element):
        for index, each_element in enumerate(self.array):
            if each_element == element:
                print(f"{each_element} for at position {index}")
                return
        print(f"{each_element} not found in array")
        self.print_array()

    def delete_an_element(self, element):
        self.array.remove(element)
        self.print_array()

    def print_array(self,):
        print(self.array)    

                       
```

```python
list = Array()
list.insert_at_end(1)
list.insert_at_end(2)
list.insert_at_end(4)
list.insert_at_end(5)
list.insert_at_end(6)

list.insert_at_position(2,3)
list.search_an_element(3)
list.delete_an_element(3)
```

## Linked List
To Tackle the problem of insertion and deletion at O(n) complexity linked list came into picture,

Linked list is a data structure where every element is store as node and node contains
1) Value
2) Address of next Node

```python
class Node:
    def __init__(self,value):
        self.next = None
        self.val = value
```

Operations of Linked List
1) Insert at position
2) Delete at position
3) Search element

```python
class Linked_List:
    def __init__(self,):
        self.head = None

    def insert_at_position(self, position, value):
        new_node = Node(value)
        if not self.head:
            self.head = new_node
        elif position == 0:
            new_node.next = self.head
            self.head = new_node
        else:
            count = 0
            temp = self.head
            while temp.next is not None and count < position-1:
                temp = temp.next
                count += 1 
            if temp.next:
                new_node.next = temp.next
                temp.next = new_node
            else:
                temp.next = new_node
        self.print_linked_list()

    def delete_at_position(self, position):
        if not self.head:
            print("List is Empty")
            return 
        elif position == 0:
            self.head = self.head.next    
        else:
            count = 0
            temp = self.head
            while temp.next is not None and count < position-1:
                temp = temp.next
                count += 1 
            if temp.next:
                temp.next = temp.next.next
            else:
                temp.next = None 
        self.print_linked_list()

    def print_linked_list(self,):
        temp = self.head
        while temp:
            print(f"{temp.val} -> ", end="")
            temp = temp.next
        print()    
```

```python
ll = Linked_List()
ll.insert_at_position(0,1)
ll.insert_at_position(1,2)
ll.insert_at_position(1,3)
ll.insert_at_position(0,4)
ll.insert_at_position(2,5)

ll.delete_at_position(0)
ll.delete_at_position(1)
ll.delete_at_position(2)
```

# Stack
Stack is linear data structure with Lifo properties "Last in first Out".

Stack can be implemented using deque 

```python
from collections import deque
s = deque()
```

Operations on stack
1) Push
2) Pop
3) Peek
4) Size

```python
class Stack:
    def __init__(self,):
        self.s = deque()

    def push(self, element):
        self.s.append(element)
        print(self.s)

    def pop(self,):
        element = self.s.pop()
        print(f"Top element of the stack is removed {element}")

    def peek(self,):
        element = self.s.pop()
        print(f"Top element of stack is {element}")
        self.s.append(element)

    def size(self,):
        print(f"Size of stack -> {len(self.s)}")
```

```python
stack = Stack()
stack.push(5)
stack.push(1)
stack.push(4)
stack.push(6)
stack.push(2)

stack.peek()
stack.pop()
stack.peek()
```

# Queue
Queue is linear data structure with Fifo properties "First in first Out".

Queue can be implemented using deque 

```python
from collections import deque
q = deque()
```

Operations on stack
1) Push
2) Pop
4) Size

```python
class Queue:
    def __init__(self,):
        self.q = deque()

    def push(self, element):
        self.q.appendleft(element)
        print(self.q)

    def pop(self,):
        element = self.q.pop()
        print(f"First element of the Queue is removed {element}")

    def size(self,):
        print(f"Size of Queue -> {len(self.q)}")
```

```python
queue = Queue()
queue.push(5)
queue.push(1)
queue.push(4)
queue.push(6)
queue.push(2)

queue.pop()
print(queue.q)
```

## Hashmap
In Python Hasmap is implemented using dictionary

Hashmap has a specific key assign to the value , like a pair. A hash function is used to map the key to value stored location

```python
hashmap = {"a":1,"b":2,"c":3}
```

Operations on HashMap
1) SetItem
2) GetItem
3) DeleteItem

```python
class HashMap:
    def __init__(self,):
        self.addr = [None]*100
        self.keys = []

    def hash_function(self,key):
        s = 0
        for each in key:
            s = s+ ord(each)
        hash = s%100
        return hash 

    def getitem(self, key):
        hash = self.hash_function(key)           
        value = self.addr[hash]
        print(f"{key}->{value}")

    def setitem(self, key, value):
        hash = self.hash_function(key)
        self.addr[hash] = value
        self.keys.append(key)

    def deleteitem(self, key):
        hash = self.hash_function(key)
        self.addr[hash] = None 
        self.keys.remove(key)

    def print_hashmap(self,):
        for key in self.keys:
            hash = self.hash_function(key)
            value = self.addr[hash]
            print(f"{key}->{value}", end = "  ")
        print()               
```

```python
d = HashMap()
d.setitem("a", 5)
d.setitem("b", 1)
d.setitem("c", 2)
d.setitem("d", 3)
d.print_hashmap()

d.getitem("c")
d.deleteitem("c")
d.getitem("c")
d.print_hashmap()
```
