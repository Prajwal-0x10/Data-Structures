# ⚡ Linear Data Structures: From Scratch in Python
> **A Comprehensive Guide to Arrays, Linked Lists, Stacks, Queues & HashMaps**

In **linear data structures**, data elements are arranged sequentially or linearly, where each element is attached to its previous and next adjacent elements.

---

## 1. Dynamic Arrays
**Core Concept (Gist):** 
Elements stored in contiguous memory locations. Offers instantaneous $O(1)$ random access by index, but inserting or deleting at arbitrary positions requires shifting elements.

### ⏱️ Complexity Summary:
| Operation | Average Case | Worst Case | Reason |
| :--- | :---: | :---: | :--- |
| **Access / Lookup** | $O(1)$ | $O(1)$ | Direct pointer arithmetic via index. |
| **Insert at End** | $O(1)$ amortized | $O(N)$ | Appends directly; occasionally resizes when capacity is exceeded. |
| **Insert at Index** | $O(N)$ | $O(N)$ | Requires shifting $N - i$ elements rightward. |
| **Delete from Index** | $O(N)$ | $O(N)$ | Requires shifting $N - i$ elements leftward. |
| **Search (Linear)** | $O(N)$ | $O(N)$ | Iterates through elements until match is found. |



````python
array = [1,2,3,4,5,6]
````

Operations on Array
1) Insert at position
2) Insert at end
3) Search an element
4) Delete an element


````python
class Array:
    def __init__(self,):
        self.array = []

    def insert_at_end(self, element):
        self.array.append(element)
        self.print_array()

    def insert_at_position(self, position, element):
        self.array.insert(position, element)
        self.print_array()

    def search_an_element(self, element):
        for index, each_element in enumerate(self.array):
            if each_element == element:
                print(f"{each_element} for at position {index}")
                return
        print(f"{each_element} not found in array")
        self.print_array()

    def delete_an_element(self, element):
        self.array.remove(element)
        self.print_array()

    def print_array(self,):
        print(self.array)    

                       
````

````python
list = Array()
list.insert_at_end(1)
list.insert_at_end(2)
list.insert_at_end(4)
list.insert_at_end(5)
list.insert_at_end(6)

list.insert_at_position(2,3)
list.search_an_element(3)
list.delete_an_element(3)
````

## 2. Singly Linked List
**Core Concept (Gist):** 
A linear collection of independent **Nodes** stored non-contiguously in memory. Each node holds data and a reference (`next`) to the succeeding node.

```
  [Head]
  ┌───────────┐      ┌───────────┐      ┌───────────┐
  │ Val │ Next│ ───> │ Val │ Next│ ───> │ Val │ Next│ ───> None
  └───────────┘      └───────────┘      └───────────┘
```

### ⚖️ Array vs. Linked List:
- **Dynamic Allocation:** No pre-allocated memory or costly array resizing.
- **Fast Insertions:** Insertion at the head is strictly $O(1)$ (no shifting).
- **Trade-off:** No direct indexing ($O(N)$ sequential traversal required) and extra memory overhead for pointer storage.



````python
class Node:
    def __init__(self,value):
        self.next = None
        self.val = value
````

Operations of Linked List
1) Insert at position
2) Delete at position
3) Search element

````python
class Linked_List:
    def __init__(self,):
        self.head = None

    def insert_at_position(self, position, value):
        new_node = Node(value)
        if not self.head:
            self.head = new_node
        elif position == 0:
            new_node.next = self.head
            self.head = new_node
        else:
            count = 0
            temp = self.head
            while temp.next is not None and count < position-1:
                temp = temp.next
                count += 1 
            if temp.next:
                new_node.next = temp.next
                temp.next = new_node
            else:
                temp.next = new_node
        self.print_linked_list()

    def delete_at_position(self, position):
        if not self.head:
            print("List is Empty")
            return 
        elif position == 0:
            self.head = self.head.next    
        else:
            count = 0
            temp = self.head
            while temp.next is not None and count < position-1:
                temp = temp.next
                count += 1 
            if temp.next:
                temp.next = temp.next.next
            else:
                temp.next = None 
        self.print_linked_list()

    def print_linked_list(self,):
        temp = self.head
        while temp:
            print(f"{temp.val} -> ", end="")
            temp = temp.next
        print()    
````

````python
ll = Linked_List()
ll.insert_at_position(0,1)
ll.insert_at_position(1,2)
ll.insert_at_position(1,3)
ll.insert_at_position(0,4)
ll.insert_at_position(2,5)

ll.delete_at_position(0)
ll.delete_at_position(1)
ll.delete_at_position(2)
````

## 3. Stack (LIFO: Last-In, First-Out)
**Core Concept (Gist):** 
A restricted linear data structure where insertions and deletions happen exclusively at one end called the **Top**. The last element added is the first one removed.

```
       |   3   |  <- Top (Pushed last, Popped first)
       |   2   |
       |___1___|
```

### ⚙️ Core Operations:
- **Push ($O(1)$):** Insert element onto the top.
- **Pop ($O(1)$):** Remove and return top element.
- **Peek ($O(1)$):** Inspect top element without removing.
- **Size / isEmpty ($O(1)$):** Check state of the stack.

> **💡 Python Pro-Tip:** Use `collections.deque` rather than `list` for stacks and queues in performance-critical code. While `list.append()` and `list.pop()` are amortized $O(1)$, `deque` provides consistent $O(1)$ memory guarantees without reallocation spikes.



````python
from collections import deque
s = deque()
````

Operations on stack
1) Push
2) Pop
3) Peek
4) Size

````python
class Stack:
    def __init__(self,):
        self.s = deque()

    def push(self, element):
        self.s.append(element)
        print(self.s)

    def pop(self,):
        element = self.s.pop()
        print(f"Top element of the stack is removed {element}")

    def peek(self,):
        element = self.s.pop()
        print(f"Top element of stack is {element}")
        self.s.append(element)

    def size(self,):
        print(f"Size of stack -> {len(self.s)}")
````

````python
stack = Stack()
stack.push(5)
stack.push(1)
stack.push(4)
stack.push(6)
stack.push(2)

stack.peek()
stack.pop()
stack.peek()
````

## 4. Queue (FIFO: First-In, First-Out)
**Core Concept (Gist):** 
A linear data structure open at both ends: elements are added at the **Rear** (Enqueue) and removed from the **Front** (Dequeue).

```
   Enqueue (Rear) ---> [ 3 | 2 | 1 ] ---> Dequeue (Front)
```

### ⚙️ Core Operations:
- **Enqueue ($O(1)$):** Add element to rear (`append`).
- **Dequeue ($O(1)$):** Remove element from front (`popleft`).
- **Peek ($O(1)$):** View front element.

> **⚠️ Critical Interview Note:** Never implement a Queue using standard Python `list` with `pop(0)`! `list.pop(0)` is $O(N)$ because every remaining element must shift left. `collections.deque.popleft()` is strictly $O(1)$ via a doubly-linked block list.



````python
from collections import deque
q = deque()
````

Operations on stack
1) Push
2) Pop
4) Size

````python
class Queue:
    def __init__(self,):
        self.q = deque()

    def push(self, element):
        self.q.appendleft(element)
        print(self.q)

    def pop(self,):
        element = self.q.pop()
        print(f"First element of the Queue is removed {element}")

    def size(self,):
        print(f"Size of Queue -> {len(self.q)}")
````

````python
queue = Queue()
queue.push(5)
queue.push(1)
queue.push(4)
queue.push(6)
queue.push(2)

queue.pop()
print(queue.q)
````

## 5. Hash Map (Dictionary)
**Core Concept (Gist):** 
An associative key-value store providing average $O(1)$ lookups, insertions, and deletions using a **Hash Function** that maps keys to array bucket indices.

```
   Key ("apple") ──[Hash Function]──> Index 3 ──> Value (42)
```

### 🛡️ Handling Collisions:
When two distinct keys hash to the same index:
1. **Separate Chaining:** Each bucket stores a linked list of entries.
2. **Open Addressing (Linear Probing):** Searches for the next available empty bucket.

### ⏱️ Time Complexity:
| Operation | Average Case | Worst Case (High Collisions) |
| :--- | :---: | :---: |
| **Insert / Set** | $O(1)$ | $O(N)$ |
| **Search / Get** | $O(1)$ | $O(N)$ |
| **Delete** | $O(1)$ | $O(N)$ |



````python
hashmap = {"a":1,"b":2,"c":3}
````

Operations on HashMap
1) SetItem
2) GetItem
3) DeleteItem

````python
class HashMap:
    def __init__(self,):
        self.addr = [None]*100
        self.keys = []

    def hash_function(self,key):
        s = 0
        for each in key:
            s = s+ ord(each)
        hash = s%100
        return hash 

    def getitem(self, key):
        hash = self.hash_function(key)           
        value = self.addr[hash]
        print(f"{key}->{value}")

    def setitem(self, key, value):
        hash = self.hash_function(key)
        self.addr[hash] = value
        self.keys.append(key)

    def deleteitem(self, key):
        hash = self.hash_function(key)
        self.addr[hash] = None 
        self.keys.remove(key)

    def print_hashmap(self,):
        for key in self.keys:
            hash = self.hash_function(key)
            value = self.addr[hash]
            print(f"{key}->{value}", end = "  ")
        print()               
````

````python
d = HashMap()
d.setitem("a", 5)
d.setitem("b", 1)
d.setitem("c", 2)
d.setitem("d", 3)
d.print_hashmap()

d.getitem("c")
d.deleteitem("c")
d.getitem("c")
d.print_hashmap()
````
